# Limier AML Model Training & Offline Feature Pre-computation Notebook

This notebook trains the Machine Learning Anomaly Detection models (Isolation Forest, XGBoost, and MetaEnsemble) on transaction history and pre-computes feature matrices and hybrid risk scores.

Running this notebook saves the trained model artifacts to `backend/models/` and pre-scored risk summaries to `backend/data/cache/`, allowing the backend server to launch instantly in **< 1 second** without re-parsing or re-building features on 200,000+ transactions every time.

```python
import os
import sys
import time
import pandas as pd
import joblib

# Ensure relative imports work from backend directory
backend_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
if backend_dir not in sys.path:
    sys.path.insert(0, backend_dir)

from src.agent.tools.anomaly_detection_tool import detect_anomalies
from src.agent.tools.risk_classification_tool import classify_risk

print(f"Backend root: {backend_dir}")
```

## 1. Load Raw Datasets

Load `transactions.csv` and `customers.csv` from `backend/data/dataset/`.

```python
data_dir = os.path.join(backend_dir, "data", "dataset")
txns_path = os.path.join(data_dir, "transactions.csv")
custs_path = os.path.join(data_dir, "customers.csv")

print(f"Loading transactions from {txns_path}...")
txns = pd.read_csv(txns_path)
txns["timestamp"] = pd.to_datetime(txns["timestamp"])

custs = pd.read_csv(custs_path)

print(f"Loaded {len(txns):,} transactions across {len(custs):,} customers.")
txns.head()
```

## 2. Feature Building & ML Anomaly Model Training

Build rolling temporal features (Categories A, B, C) and train:
1. **Isolation Forest** (Unsupervised anomaly scoring)
2. **XGBoost** (Supervised gradient boosted trees with SHAP feature attributions)
3. **MetaEnsemble** (Calibrated Platt logistic meta-model)

```python
models_dir = os.path.join(backend_dir, "models")
os.makedirs(models_dir, exist_ok=True)

# Clear existing models to force fresh fit
for m in ["iso_scorer.pkl", "xgb_scorer.pkl", "meta_ensemble.pkl"]:
    p = os.path.join(models_dir, m)
    if os.path.exists(p):
        os.remove(p)

start_t = time.time()
print("Building AML features and fitting ML models...")
features_df, iso_scorer, xgb_scorer, meta_ensemble = detect_anomalies(txns)
print(f"Feature building and ML model fitting completed in {time.time() - start_t:.2f} seconds.")
```

## 3. Risk Classification & Rule Hybridization

Combine ML anomaly scores with deterministic AML typologies (Structuring, Velocity, Rapid Cashout, Dormancy Spikes) and compute SHAP explanations.

```python
start_t = time.time()
print("Classifying risk levels and SHAP feature impacts...")
risk_summary_df = classify_risk(txns, features_df)
print(f"Risk classification completed in {time.time() - start_t:.2f} seconds.")

print("Risk breakdown:")
print(risk_summary_df["risk_level"].value_counts())
risk_summary_df.head()
```

## 4. Save Pre-computed Cache to Disk

Persist `risk_summary_df` to disk so the FastAPI server loads instantly.

```python
cache_dir = os.path.join(backend_dir, "data", "cache")
os.makedirs(cache_dir, exist_ok=True)

risk_cache_path = os.path.join(cache_dir, "risk_summary_df.pkl")
features_cache_path = os.path.join(cache_dir, "features_df.pkl")

joblib.dump(risk_summary_df, risk_cache_path)
joblib.dump(features_df, features_cache_path)

print(f"✅ Successfully saved pre-scored dataset cache to:\n  - {risk_cache_path}\n  - {features_cache_path}")
print("Backend startup will now be < 1 second!")
```
